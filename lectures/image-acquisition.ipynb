# Image Acquisition and Encoding

In this module we examine the image formation and the image acquisition process as well as the image encoding.

## 3D image acquisition

There are two different kinds of 3D images. Surface images and volume images. Surface images contain only the information about the surface of an object, no information about the inside of the object is acquired. Surface images can be created by sampling information on the surface of the object. Volume images in contrast, also contain information from inside the object, a sampling of the whole object is therefore necessary.

Technologies to acquire surface images are:

1. Laser Scanner (LiDAR)<br>
LiDAR stands for "Light Detection and Ranging". Laser pulses are emitted towards a target and the reflections are captured by a detector. The distance to the object is calculated from the time between the emission of the pulse and the detection of the reflection. A point cloud of the surfaces of objects is constructed by moving the laser and/or changing the angle.
2. Structured-light 3D scanner<br>
projects a pattern (stripes or a grid) onto a surface and constructs the 3D object from the deformations of the pattern.
3. Stereophotogrammetry<br>
Uses the images of multiple cameras at different positions to reconstruct the 3D surface.
4. Photometric 3D SEM<br>
In scanning electron microscopy a 3d surface can be calculated from two images taken at different angles.

To create volume images we need to obtain information from the inside of the object. That means we either have to cut the object into thin sections or the objects need to let light or other waves penetrate. If necessary tissue clearing can be used to make samples more transparent.

Technologies for volume imaging are:

1. Thin sectioning<br>
The samples are embedded into a hardening medium, for example Paraffin, or frozen and cut into thin sections (in the nanometer to micrometer range).
2. Optical Sectioning<br>
The volume is reconstructed by moving the focal plane of a microscope across a specimen along the x-axis and taking an image at each z-position. The microscope needs to be able to exclude out-of-focus light to some degree.
3. CT scan<br>
X-ray measurements from different angles are used to produce a stack of tomographic, i.e. cross-sectional, images.
4. Magnetic Resonance Imaging (MRI)<br>
 MRI creates 3D images by manipulating the magnetic properties of hydrogen atoms, using varying magnetic fields to "map" their locations

In the next sections, we will have a closer look at thin sectioning and optical sectioning.

### Thin Sectioning

Thin sectioning is a fundamental technique in histology and microscopy used to transform a thick, opaque specimen into a series of ultra-thin, transparent slices that can be imaged.

The sample must be fixed. Live imaging is of course not possible. The fixation stabilizes the biological structures and prevents decay. To enhance the contrast the sample is often stained. Chemical staining will give regions with specific chemical properties a distinct color. The Hematoxylin and eosin stain (H&E) will for example give the nuclei a purplish-blue color and stains the cytoplasm and extracellular matrix pink.

Soft tissue can be damaged or deformed when cut into slices. It can be necessary to freeze the sample or to embed it into a medium, like for example paraffin. A microtome or ultramicrotome is used to cut the sample into thin slices. The microtome allows to adjust the thickness and uses a steel, glass or diamant blade. Parafin embedded sections usually have a thickness between 3-50µm. Semi-thin sections that can be used for electron and light microscopy can produced by embedding the sample in acrylic or epoxy resins. They have a thickness between 0.5-2µm. To produce ultra-thin sections for transmission electron microscopy, the sample is embedded in epoxy resins and cut using an ultramicrotome. The thickness is between 50-100nm.
